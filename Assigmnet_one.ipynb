# Question one 


```python
text = input("Enter letters and numbers: ")

letters = []
numbers = []

for character in text:
    if character.isalpha():
        letters.append(character)
    elif character.isdecimal():
        numbers.append(int(character))

word = "".join(letters)
total = sum(numbers)

print("Word:", word)
print("Sum:", total)
```

# Question 2

```python
password = input("Enter a password: ")

special_characters = "!@#$%^&*()-_+=[]{};:,.<>?/\\|~`'\""

has_uppercase = False
has_special = False

for character in password:
    if character.isupper():
        has_uppercase = True

    if character in special_characters:
        has_special = True

if 8 <= len(password) <= 16 and has_uppercase and has_special:
    print("Password is Secure.")
else:
    print("Password is not Secure.")
```

# Question 3

```python
price = float(input("Enter the total price: "))
member = input("Are you a member? (yes/no): ").strip().lower()

if price < 50:
    discount = 0
elif price <= 100:
    discount = 0.10
else:
    discount = 0.20

final_price = price * (1 - discount)

if member == "yes":
    final_price = final_price * 0.95

print(f"Final Total: ${final_price:.2f}")
```

# Question 4

```python
list1 = input("Enter the first list of names, comma-separated: ").split(",")
list2 = input("Enter the second list of names, comma-separated: ").split(",")

for i in range(len(list1)):
    list1[i] = list1[i].strip()

for i in range(len(list2)):
    list2[i] = list2[i].strip()

common_names = []

for name in list1:
    if name != "" and name in list2 and name not in common_names:
        common_names.append(name)

print("Common names:", common_names)
```

# Question 7

```python
number = int(input("Enter a positive multi-digit number: "))
number = abs(number)

while number >= 10:
    total = 0

    for digit in str(number):
        total = total + int(digit)

    number = total

print("Digital Root:", number)
```

# Question 9

```python
n = int(input("Enter a positive integer: "))

if n < 1:
    print("Please enter a number greater than or equal to 1.")
else:
    factorial = 1
    total = 0

    for number in range(1, n + 1):
        factorial *= number
        total += number

    print(f"Factorial is {factorial}, Sum is {total}.")

    if factorial % total == 0:
        print(f"{factorial} is divisible by {total}.")
    else:
        print(f"{factorial} is not divisible by {total}.")
```

# Question 10 

```python
store = {
    "apple": 0.5,
    "bread": 2.0,
    "milk": 1.5
}

cart = []

while True:
    action = input("Choose Add, Remove, or Done: ").strip().lower()

    if action == "done":
        break

    elif action == "add":
        item = input("Enter the item to add: ").strip().lower()

        if item in store:
            cart.append(item)
            print(item, "added.")
        else:
            print("Error: This item is not in the store.")

    elif action == "remove":
        item = input("Enter the item to remove: ").strip().lower()

        if item in cart:
            cart.remove(item)
            print(item, "removed.")
        else:
            print("This item is not in your cart.")

    else:
        print("Invalid action. Choose Add, Remove, or Done.")

total = 0

for item in cart:
    total += store[item]

print(f"Total Bill: ${total:.2f}")
```

# Question 13 

```python
numbers = [1, 5, 2, 6, 3]

peaks = []

for i in range(1, len(numbers) - 1):
    if numbers[i] > numbers[i - 1] and numbers[i] > numbers[i + 1]:
        peaks.append(numbers[i])

print("Peaks:", peaks)
```

# Question 14 

```python
original = {
    "A": 1,
    "B": 2,
    "C": 1
}

flipped = {}

for key, value in original.items():
    if value not in flipped:
        flipped[value] = []

    flipped[value].append(key)

print(flipped)
```

```python
text = input("Enter a string: ").lower()

vowels = {}

for i in range(len(text)):
    character = text[i]

    if character in "aeiou":
        if character not in vowels:
            vowels[character] = []

        vowels[character].append(i)

print(vowels)
```
